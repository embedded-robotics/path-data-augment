```python
import os
```

```python
data_dir = "/data/mn27889/path-data-augment/taming-transformers/data"
source_image_dir = os.path.join(data_dir, "path_vqa_images")
```

```python
splits = ["train", "validation", "test"]
for split in splits:
    split_dir = os.path.join(source_image_dir, split)
    split_dir_list = os.listdir(split_dir)
    split_image_path = [os.path.join(split_dir, image) for image in split_dir_list]
    with open(os.path.join(data_dir, f"path_vqa_{split}.txt"), "w") as f:
        f.write("\n".join(split_image_path))
```
